```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Fine tuned model by my team
tokenizer1 = AutoTokenizer.from_pretrained("SanketAI/FLAN-T5_instruct-mistral7b")
model1 = AutoModelForSeq2SeqLM.from_pretrained("SanketAI/FLAN-T5_instruct-mistral7b")

# Base model
tokenizer2 = AutoTokenizer.from_pretrained("google/flan-t5-base")
model2 = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-base")

# Define input prompt
input_text = "Translate the following English text to French: 'Hello, how are you?'"

# Tokenize the input text for both models
inputs1 = tokenizer1(input_text, return_tensors="pt", padding=True, truncation=True)
inputs2 = tokenizer2(input_text, return_tensors="pt", padding=True, truncation=True)

# Generate output from both models
outputs1 = model1.generate(inputs1["input_ids"], max_length=50, num_beams=5, early_stopping=True)
outputs2 = model2.generate(inputs2["input_ids"], max_length=50, num_beams=5, early_stopping=True)

# Decode the outputs from both models
generated_text1 = tokenizer1.decode(outputs1[0], skip_special_tokens=True)
generated_text2 = tokenizer2.decode(outputs2[0], skip_special_tokens=True)

print(input_text)
# Print the results for comparison
print("Output from SanketAI/FLAN-T5_instruct-mistral7b:")
print(generated_text1)
print("\nOutput from google/flan-t5-base:")
print(generated_text2)
```
